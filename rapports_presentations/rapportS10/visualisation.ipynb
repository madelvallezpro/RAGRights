## Visualisation des matrices de sensibilité

```python

from mechir.plotting import plot_components

VIEW_STYLES = {
    "base" : {
        "cmap" :"RdBu_r",
        "vmin" : -1,
        "vmax" : 1,
    },
    "base_adapt" : {
        "cmap" :"RdBu_r",
        "vmin" : -1,
        "vmax" : 1,
    },
    "base_extr" : {
        "cmap" :"RdBu_r",
        "vmin" : -50,
        "vmax" : 50,
    },
    "base_abs" : {
        "cmap" : "Reds",
        "vmin" : 0,
        "vmax" : 1,
    },
    "base_resc" : {
        "cmap" : "RdBu_r",
        "vmin" : -1,
        "vmax" : 1,
    },
    "std" : {
        "cmap" : "magma_r",
        "vmin" : 0,
        "vmax" : 1,
    },
    "std_adapt" : {
        "cmap" : "magma_r",
        "vmin" : 0,
        "vmax" : 1,
    },
    "std_max" : {
        "cmap" : "magma_r",
        "vmin" : 0,
        "vmax" : 50,
    },
    "std_resc" : {
        "cmap" : "magma_r",
        "vmin" : 0,
        "vmax" : 1,
    },
    "norm" : {
        "cmap" : "RdBu_r",
        "vmin" : -1,
        "vmax" : 1,
    },
    "std_norm" : {
        "cmap" : "magma_r",
        "vmin" : 0,
        "vmax" : 1,
    },
    "abs" : {
        "cmap" : "magma_r",
        "vmin" : 0,
        "vmax" : 50,
    },
}

def plot_components_V2(
    data,  # shape: (num_layers, num_heads) or (num_layers, num_heads + 1) if include_mlp=True
    save_path=None,
    title="Component Patching Results",
    include_mlp=False,
    view_style = "base"
):

    
    data = data.astype(float)
    plt.figure(figsize=(10, 6))

    printed_data = data
    view_params = VIEW_STYLES[view_style].copy()
    if "norm" in view_style:
        # printed_data = printed_data / np.linalg.norm(data)
        printed_data = (printed_data - printed_data.mean()) /printed_data.std()
    if "abs" in view_style:
        printed_data = printed_data.absolute()
    if "extr" in view_style:
        view_params["vmax"] = (abs(printed_data.max())+abs(printed_data.min())) /2
        view_params["vmin"] = - view_params["vmax"]
    if "max" in view_style:
        view_params["vmax"] = (abs(printed_data.max())+abs(printed_data.min())) /2
        view_params["vmin"] = 0
    if "adapt" in view_style:
        if (printed_data > 20).any() :
            view_params["vmax"] = 50
        else:
            view_params["vmax"] = 1
        view_params["vmin"] = (VIEW_STYLES[view_style]["vmin"] * view_params["vmax"]) # si vmin est déjà à 0 ça reste 0, si vmin est -1, on obtient -vmax
    if "resc" in view_style:
        printed_data = (np.sign(printed_data) / np.absolute(printed_data).max()) * np.absolute(printed_data)

    ax = sns.heatmap(
        printed_data,
        cmap=view_params["cmap"], 
        vmin=view_params["vmin"],
        vmax=view_params["vmax"],
        xticklabels=True,
        yticklabels=True,
        fmt=".2f",
        annot=True,
        annot_kws={"size": 8},
    )

    if include_mlp:
        new_labels = list(map(str, range(data.shape[1] - 1)))
        new_labels.append("MLP")
        ax.set_xticklabels(new_labels)

    plt.title(title)
    plt.xlabel("Head")
    plt.ylabel("Layer")

    if save_path:
        plt.savefig(save_path)
        plt.close()

    else:
        plt.show()

```

## Matrices des noeuds sensibles

```python
with open("exp5/elephants/perts_valuesV2.json", "r") as f:
    el = json.load(f)

with open("exp5/elephants-roses/perts_valuesV2.json", "r") as f:
    er = json.load(f)

with open("exp5/synonymes/perts_valuesV2.json", "r") as f :
    syn = json.load(f)

with open("exp5/polysemie/perts_valuesV2.json", "r") as f:
    data_ps = json.load(f)

with open("exp5/plus/perts_values.json", "r") as f:
    data_plus = json.load(f)

ps = data_plus + data_ps[:3] + data_ps[4:]
```

```python
def sensibles(mean, std):
    resc_mean = (np.sign(mean) / np.absolute(mean).max()) * np.absolute(mean)
    resc_std = (np.sign(std) / np.absolute(std).max()) * np.absolute(std)
    return ((np.abs(mean) > 0.1) & (np.abs(resc_mean) > 0.5)) | ((np.abs(std) > 0.1) & (np.abs(resc_std) > 0.5))
```

```python
mat_noeuds_sensibles = np.zeros((12,12))
a_traiter = syn
for p in a_traiter:
    mean = np.asarray(p["results"]["patching_mean"])
    std = np.asarray(p["results"]["patching_std"])

    carte_sensible_p = sensibles(mean, std)
    mat_noeuds_sensibles +=carte_sensible_p

ax = sns.heatmap(
    mat_noeuds_sensibles,
    cmap="Reds", 
    vmin=0,
    vmax=len(a_traiter),
    xticklabels=True,
    yticklabels=True,
    annot=True,
    annot_kws={"size": 8},
)
plt.title("Noeuds sensibles aux perturbations de niveau 1")
plt.xlabel("Head")
plt.ylabel("Layer")
```
